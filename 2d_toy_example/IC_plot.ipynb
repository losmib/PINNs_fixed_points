```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
fig, ax = plt.subplots()

intial_conditions = [(0.01, 0.01),
                    (0.1, 0.1),
                    (0.5, 0.5),
                    (1.01, 1.01),
                    (1.5, 1.5),
                    (1.5, 0.01),
                    (2.0, 1.75),
            ]

ICs1, ICs2 = [], []

for IC1, IC2 in intial_conditions:
    ICs1.append(IC1)
    ICs2.append(IC2)

lim_low, lim_high = -1, 3.2
dx = .01

f = lambda x,y: 3*x - x**2 - 2*x*y
g = lambda x,y: 2*y -x*y - y**2
X, Y = np.meshgrid(np.arange(lim_low, lim_high, dx), np.arange(lim_low, lim_high, dx))
F = f(X, Y)
G = g(X, Y)
magnitude = np.sqrt(F**2 + G**2)

#ax.quiver(X, Y, F, G, scale=50)
#ax.streamplot(X, Y, F, G, color=magnitude, cmap='jet')
ax.streamplot(X, Y, F, G, color='0.8', zorder=0)

ax.scatter(ICs1, ICs2, c='black', label='ICs')
unstable_fps = [0, 1], [0, 1]
stable_fps = [0, 3], [2, 0]

ax.scatter(unstable_fps[0], unstable_fps[1], c='red', label="unstable fp")
ax.scatter(stable_fps[0], stable_fps[1], c='green', label="stable fp")

ax.set_xlabel("x")
ax.set_ylabel("y")
plt.legend()
plt.tight_layout()
plt.show()
```
